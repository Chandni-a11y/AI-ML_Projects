```python
import numpy as np
import pandas as  pd
import xgboost as xgb
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier,VotingClassifier
from sklearn.metrics import accuracy_score, recall_score,classification_report,r2_score,precision_score
```

```python
nov = pd.read_csv(r"C:\Users\This PC\Downloads\novagen_dataset.csv")
nov.head()
```

```python
x = nov.drop(columns = ["Target"])
y = nov["Target"]

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state = 42,stratify = y)

ss = StandardScaler()
x_train_scaled = ss.fit_transform(x_train)
x_test_scaled = ss.transform(x_test)
```

# Base Model

```python
# Logistic Regression
lr = LogisticRegression(penalty = "l2",max_iter = 1000)
lr.fit(x_train_scaled,y_train)

y_pred = lr.predict(x_test_scaled)
print("accuracy:",accuracy_score(y_test,y_pred))
print("recall:",recall_score(y_test,y_pred))
```

```python
# Random Forest Classifier
rf = RandomForestClassifier(max_depth = 3,n_estimators = 100,random_state = 42)
rf.fit(x_train_scaled,y_train)

y_pred_rf = rf.predict(x_test_scaled)

print("accuracy:",accuracy_score(y_test,y_pred_rf))
print("recall:",recall_score(y_test,y_pred_rf))
print("Cr:",classification_report(y_test,y_pred_rf))
```

```python
# Knn Classifier

knn = KNeighborsClassifier(n_neighbors = 5,metric = 'euclidean')
knn.fit(x_train_scaled,y_train)

y_pred_knn = knn.predict(x_test_scaled)
print("accuracy:",accuracy_score(y_test,y_pred_knn))
print("recall:",recall_score(y_test,y_pred_knn))
print("cr:",classification_report(y_test,y_pred_knn))
```

```python
# Gradient Classifier
gb = GradientBoostingClassifier(n_estimators = 150,learning_rate = 0.1,max_depth = 3,random_state = 42)
gb.fit(x_train_scaled,y_train)

y_pred_gb = gb.predict(x_test_scaled)
print("accuracy:",accuracy_score(y_test,y_pred_gb))
print("cr:",classification_report(y_test,y_pred_gb))
```

```python
# Voting Classifier

vc = VotingClassifier(estimators = [("lr",lr),("knn",knn),("rf",rf)],voting = "soft")
vc.fit(x_train_scaled,y_train)

y_pred_vc = vc.predict(x_test_scaled)

print("acc:",accuracy_score(y_test,y_pred_vc))
print("recall:",recall_score(y_test,y_pred_vc))
print("cr:",classification_report(y_test,y_pred_vc))
```
